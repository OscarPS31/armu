```python
df = pd.read_csv('raw_data/recipes_ingredients.csv')

def parse_list_safe(x):
    try:
        return ast.literal_eval(x)
    except (ValueError, SyntaxError):
        matches = re.findall(r'"([^"]*)"|\\(.*?)\\', x)
        return [a if a else b for a, b in matches]

df['ingredients'] = df['ingredients'].apply(parse_list_safe)
df['ingredients_raw'] = df['ingredients_raw'].fillna('[]').apply(parse_list_safe)

# Donde ingredients está vacío, usa ingredients_raw como respaldo (limpiando espacios)
mask_empty = df['ingredients'].apply(len) == 0
df.loc[mask_empty, 'ingredients'] = df.loc[mask_empty, 'ingredients_raw'].apply(
    lambda lst: [item.strip() for item in lst]
)

# Conteo final
(df['ingredients'].apply(len) == 0).sum()
```

```python
df = df[df['ingredients'].apply(len) > 0].reset_index(drop=True)
df.shape
```

```python
df['description'] = df['description'].fillna('')
df['servings'] = df['servings'].fillna(df['servings'].median())

df.isnull().sum()
```

```python
restricciones = {
    'is_vegan': 'vegan',
    'is_vegetarian': 'vegetarian',
    'is_gluten_free': 'gluten-free',
    'is_dairy_free': 'dairy-free',
    'is_lactose_free': 'lactose',
    'is_nut_free': 'nut-free',
    'is_egg_free': 'egg-free',
}

for columna, tag in restricciones.items():
    df[columna] = df['tags'].apply(lambda t: tag in t)

tiempos = {
    'quick_60min': '60-minutes-or-less',
    'quick_30min': '30-minutes-or-less',
    'quick_15min': '15-minutes-or-less',
}

for columna, tag in tiempos.items():
    df[columna] = df['tags'].apply(lambda t: tag in t)


df[list(restricciones.keys()) + list(tiempos.keys())].sum()
```

```python
df.to_csv('raw_data/recipes_clean.csv', index=False)
```
