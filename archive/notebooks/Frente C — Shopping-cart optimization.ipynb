```python
#Preparar ingredients_precios.csv inventado con 10-15 ingredientes y precios - Completado
prices_df = pd.read_csv('raw_data/ingredient_prices_v1.csv')
prices_df = prices_df.drop_duplicates(
    subset=["ingredient", "store"]
).reset_index(drop=True)
prices_df
```

```python
ingredients_raw_sample = pd.read_csv('raw_data/recipe_ingredients_raw_sample.csv')
ingredients_raw_sample
```

```python
def get_ingredients_from_menu(
    menu_semanal,
    recipes_df
):
    recipe_ids = [
        recipe_id
        for daily_ids in menu_semanal.values()
        for recipe_id in daily_ids
    ]

    ingredients_list = []

    for recipe_id in recipe_ids:
        recipe_ingredients = (
            recipes_df.loc[
                recipes_df["id"] == recipe_id,
                ["ingredient", "quantity"]
            ]
            .dropna(subset=["ingredient"])
            .to_dict(orient="records")
        )

        ingredients_list.extend(recipe_ingredients)

    return ingredients_list
```

```python
ingredients_list = get_ingredients_from_menu(
    menu_sample,
    ingredients_raw_sample
)

ingredients_list
```

```python
def aggregate_ingredients(ingredients_list):
    ingredients_df = pd.DataFrame(ingredients_list)

    ingredients_df["ingredient"] = (
        ingredients_df["ingredient"]
        .str.lower()
        .str.strip()
    )

    ingredients_df["quantity"] = pd.to_numeric(
        ingredients_df["quantity"],
        errors="coerce"
    )

    aggregated_ingredients = (
        ingredients_df
        .dropna(subset=["ingredient", "quantity"])
        .groupby(
            "ingredient",
            as_index=False
        )
        .agg(
            quantity=("quantity", "sum"),
            recipe_appearances=("ingredient", "size")
        )
    )

    return aggregated_ingredients
```

```python
aggregated_ingredients = aggregate_ingredients(
    ingredients_list
)

aggregated_ingredients
```

```python
selected_store = "Chedraui"
store_prices = prices_df[
    prices_df["store"] == selected_store
].copy()

store_prices
```

```python
cart = aggregated_ingredients.merge(
    store_prices[
        [
            "ingredient",
            "profeco_category",
            "price",
            "unit",
            "store"
        ]
    ],
    on="ingredient",
    how="left",
    validate="one_to_one"
)

cart
```

```python
cart["cost"] = (
    cart["quantity"]
    * cart["price"]
).round(2)
cart
```

```python
total_cost = round(
    float(cart["cost"].sum()),
    2
)

print("Total cost:", total_cost)
```

```python
missing_ingredients = (
    cart.loc[
        cart["price"].isna(),
        "ingredient"
    ]
    .tolist()
)

print(
    "Ingredients without price:",
    missing_ingredients
)
```
