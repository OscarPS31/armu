```python
import pandas as pd, ast, re

df_ale = pd.read_csv("data/recipes_51k_eng.csv")

def leer_pasos(s):
    s = str(s).strip()
    # 1) intento normal
    try:
        r = ast.literal_eval(s)
        if isinstance(r, list):
            return [str(x).strip() for x in r if str(x).strip()]
    except Exception:
        pass
    # 2) rescate: separo por el patrón real entre pasos
    if s.startswith("["): s = s[1:]
    if s.endswith("]"): s = s[:-1]
    partes = re.split(r'",\s*"', s)
    return [p.strip().strip('"').strip() for p in partes if p.strip().strip('"').strip()]

df_ale["steps_ok"] = df_ale["steps"].apply(leer_pasos)

vacios = (df_ale["steps_ok"].apply(len) == 0).sum()
print("Recetas con pasos recuperados:", (df_ale["steps_ok"].apply(len) > 0).sum())
print("Recetas que quedan sin pasos :", vacios)
print()
print("Ejemplo recuperado:")
for p in df_ale.loc[0, "steps_ok"][:3]:
    print(" -", p)
```
